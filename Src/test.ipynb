```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the data
df = pd.read_csv('csv/Législatives/Legislatives_T1_1958.csv')
```

```python
# Display the first 5 rows
print(df.head())
```

```python
# plot the data
plt.figure(figsize=(10, 6))
sns.set(style="dark")
sns.countplot(x='Votants', data=df)
plt.xticks(rotation=45)
plt.show()
```

```python
# info about the data
print(df.info())
```

```python
# describe the data
print(df.describe())
```

```python
# histplot
plt.figure(figsize=(10, 6))
sns.histplot(df['Votants'], bins=30, kde=True)
plt.show()

```

```python
# scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Inscrits', y='Votants', data=df)
plt.show()

```

```python
# countplot
plt.figure(figsize=(10, 6))
sns.countplot(x='Votants', data=df)
plt.xticks(rotation=45)
plt.show()

# pairplot
sns.pairplot(df)
plt.show()



```
